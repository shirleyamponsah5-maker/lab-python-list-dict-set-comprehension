## Reviewing my previous exercise to use comprehension

Below, each section shows **Before** (my original code from the flow-control lab) and **After** (the same logic rewritten with a comprehension), following the lab's hints: initializing inventory, getting customer orders, calculating total price, updating inventory, and printing the updated inventory.

### 1. Initialize inventory

**Before** - a `for` loop that builds `inventory` one key at a time:

products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}
for product in products:
    quantity = int(input(f"Enter quantity for {product}: "))
    inventory[product] = quantity

**After** - the same result as a **dict comprehension**, wrapped in a reusable function:

````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def initialize_inventory(products):
    inventory = {
        product: int(input(f"Enter quantity for {product}: "))
        for product in products
    }
    return inventory

inventory = initialize_inventory(products)
````

### 2. Get customer orders

**Before** - a `while True` loop with manual validation (checks the product is real, asks "add another?" until the user says no):
```
customer_orders = set()
while True:
    order = input("Enter the name of a product to order: ").lower()
    if order in products:
        customer_orders.add(order)
    else:
        print("invalid product. Choose from:", products)
        continue
    another_product = input("Do you want to add another_product? (yes/no: ").lower()
    if another_product == "no":
        break
```

**After** — the lab asks us to simplify this to "prompt for the number of orders, then gather names with comprehension". A **set comprehension** can't easily retry on invalid input the way the `while` loop did, so this version trades away the invalid-product check in exchange for the more compact, comprehension-driven flow the lab is asking for:

````python
def get_customer_orders():
    num_orders = int(input("Enter the number of customer orders: "))
    customer_orders = {
        input("Enter the name of a product to order: ").lower()
        for _ in range(num_orders)
    }
    return customer_orders

customer_orders = get_customer_orders()
````

### Printing customer orders (fixing a small bug along the way)

**Before** - this had a small bug: it loops over `customer_orders` but prints the *whole set* on every iteration instead of the individual `product`, so a 3-item set printed the same full set 3 times:
```python
print("\nCustomer Orders:")
for product in customer_orders:
     print(customer_orders)   # bug: should print `product`, not `customer_orders`
```

**After** - printing each product once, cleanly:

````python
print("\nCustomer Orders:")
for product in customer_orders:
    print(product)
````

### 3. Calculate total price (new function)

This function wasn't in the original exercise - it's new. For each product in `customer_orders`, we ask for its price and use a **dict comprehension** to build `{product: price}`, then sum the values (each product is ordered once, so a simple sum is correct).

````python
def calculate_total_price(customer_orders):
    prices = {
        product: float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    }
    total_price = sum(prices.values())
    return total_price
````

### Order statistics

This part didn't need a comprehension , it's already a couple of simple calculations so it stays as it is.

````python
total_products_ordered = len(customer_orders)
percentage_product_ordered = (total_products_ordered / len(products)) * 100
order_status = (total_products_ordered, percentage_product_ordered)

print("\nOrder Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")
````

### 4. Update inventory (remove zero-quantity items)

**Before** - a `for` loop that decrements quantity for ordered products, but never removes anything that hits zero:
```python
for product in customer_orders:
    if product in inventory:
        inventory[product] -= 1
```

**After** - wrapped in a function, using two **dict comprehensions**: one to decrement ordered products, and a second to filter out anything that dropped to 0:

````python
def update_inventory(customer_orders, inventory):
    inventory = {
        product: (quantity - 1 if product in customer_orders else quantity)
        for product, quantity in inventory.items()
    }
    inventory = {
        product: quantity
        for product, quantity in inventory.items()
        if quantity > 0
    }
    return inventory

inventory = update_inventory(customer_orders, inventory)
````

### 5. Print updated inventory and total price

The updated-inventory print loop stays a `for` loop (printing key/value pairs isn't something a comprehension improves it's a side effect, not a value we're building), and finally we call the new `calculate_total_price` function and print the result:

````python
print("\nUpdated Inventory:")
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")

total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price}")
````
